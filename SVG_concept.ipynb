```python
def create_rectangle_svg(filename):
    svg_content = """<?xml version="1.0" encoding="UTF-8"?>
    <svg xmlns="http://www.w3.org/2000/svg"
        width="400"
        height="300">

        <rect
            x="50"
            y="50"
            width="300"
            height="200"
            fill="lightblue"
            stroke="black"
            stroke-width="3" />

    </svg>
    """

    with open(filename, "w") as f:
        f.write(svg_content)

create_rectangle_svg("rectangle.svg")
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

fig, ax = plt.subplots(figsize=(8, 4))

# Rectangle(
#     (x, y),   <- lower-left corner of rectangle
#     width,    <- rectangle width in plot units
#     height,   <- rectangle height in plot units
# )

# rect1 = Rectangle(
#     (1, 1),            # lower-left corner at x=1, y=1
#     4,                 # width = 4 units (x = 1 to 5)
#     2,                 # height = 2 units (y = 1 to 3)
#     facecolor="lightblue",
#     edgecolor="black",
#     linewidth=2,
# )

x_start_point = 1
y_start_point = 1
width = 4
height = 2
gap_btwn_rects = 1

num_rects = 3

for i in range(0, num_rects):

    x = x_start_point + i * (width + gap_btwn_rects)

    rect = Rectangle(
        (x, y_start_point),  # lower-left corner at x=1, y=1
        width,                           # width = 4 units (x = 1 to 5)
        height,                          # height = 2 units (y = 1 to 3)
        facecolor="lightblue",
        edgecolor="black",
        linewidth=2,
    )
    
    ax.add_patch(rect) # draw this shape on the axes

# Text placed at x=3, y=2
# (the center of the rectangle)
# ax.text(
#     3,
#     2,
#     "Valve Station 1",
#     fontsize=12,
#     ha="center",       # horizontal alignment relative to x
#     va="center",       # vertical alignment relative to y
# )

# Plot boundary limits
# x-axis spans from 0 to 6
ax.set_xlim(0, (x_start_point + num_rects * (width + gap_btwn_rects) + 1))

# y-axis spans from 0 to 4
ax.set_ylim(0, 4)

# Makes 1 unit in x equal to 1 unit in y
# Prevents distortion of shapes
ax.set_aspect("equal")

# Hides axes, ticks, and labels
ax.axis("off")

plt.savefig("drawing.pdf", bbox_inches="tight")
```
